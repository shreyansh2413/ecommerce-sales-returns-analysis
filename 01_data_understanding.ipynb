```python
# Customers whose first_order_date is missing
missing_first_order = customers[
    customers["first_order_date"].isna()
]

# Among them, find the customer IDs who actually have orders
problem_customer_ids = missing_first_order[
    missing_first_order["customer_id"].isin(
        orders["customer_id"]
    )
]["customer_id"]

# Check those customers' orders
problem_orders = orders[
    orders["customer_id"].isin(problem_customer_ids)
]

print(
    problem_orders[
        ["customer_id", "order_id", "order_date",
         "order_status", "order_total_usd"]
    ].sort_values(["customer_id", "order_date"])
)
```

```python
# Orders → Customers
missing_customers = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

# Order Items → Orders
missing_orders = order_items[
    ~order_items["order_id"].isin(orders["order_id"])
]

# Order Items → Products
missing_products = order_items[
    ~order_items["product_id"].isin(products["product_id"])
]

# Returns → Order Items
missing_return_items = returns[
    ~returns["order_item_id"].isin(order_items["order_item_id"])
]

print("Orders with missing customer:", len(missing_customers))
print("Order items with missing order:", len(missing_orders))
print("Order items with missing product:", len(missing_products))
print("Returns with missing order item:", len(missing_return_items))
```

```python
calculated_line_total = (
    order_items["quantity"] * order_items["unit_price_local"]
    - order_items["line_discount_local"]
)

difference = (
    calculated_line_total
    - order_items["line_total_local"]
).abs()

print("Maximum difference:", difference.max())
print("Rows with mismatch:", (difference > 0.01).sum())
```
